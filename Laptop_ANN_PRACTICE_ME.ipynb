```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from tensorflow.keras.models import Sequential
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.layers import Dense,Dropout,BatchNormalization
```

```python
df=pd.read_csv(r'C:\Users\amank\Downloads\DEEP LEARNING ANN\Laptop_data.csv')
df.head()
```

```python
#                      DROP UNWANTED COLUMNS
```

```python
df.drop(columns=['Company','TypeName','Os','Gpu_brand','TouchScreen'],inplace=True)
df.head()
```

```python
label=LabelEncoder()
```

```python
df.isnull().sum()
```

```python
df.drop_duplicates()
```

```python
df['Cpu brand']=label.fit_transform(df['Cpu_brand'])
```

```python
df.head()
```

```python
df.dropna()
```

```python
x=df.drop(columns=['Cpu_brand','Price'])
y=df['Price']
```

```python
x.shape
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
model=Sequential()
```

```python
model.add(Dense(35,activation='relu',input_dim=7))
model.add(Dense(32,activation='relu'))
model.add(Dense(32,activation='sigmoid'))
```

```python
model.summary()
```

```python
#           USE SPARSE_CATEGORICAL_CROSSENTRPY (when the columsn miss match)
```

```python
model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
model.fit(x_train,y_train,batch_size=32,epochs=10)
```
