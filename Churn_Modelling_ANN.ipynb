```python
import pandas as pd
from sklearn.model_selection import train_test_split
from tensorflow.keras.models import Sequential   # MAINTAINS THE SEQUENCE OF THE MODEL.
from tensorflow.keras.layers import Dense   # 1D LAYER.
from tensorflow.keras.layers import Dropout,BatchNormalization  # PREVENTS OVERFITTING.
from tensorflow.keras.callbacks import EarlyStopping  # PREVENTS OVERFITTING.
from sklearn.preprocessing import LabelEncoder
```

```python
#                            ANN (Artificail Nural Networks)
```

```python
df=pd.read_csv(r'C:\Users\amank\Downloads\DEEP LEARNING ANN\Churn_Modelling.csv')
df.head()
```

```python
#                                 DROP USELESS COLUMNS
```

```python
df.drop(columns=['RowNumber','CustomerId','Surname','Geography'],inplace=True)
df.head()
```

```python
#                      HANDLING CATEGORICAL DATA
```

```python
label=LabelEncoder()
```

```python
df['encoded_gender']=label.fit_transform(df['Gender'])
```

```python
df.drop(columns=['Gender'],inplace=True)
```

```python
df.head()
```

```python
x=df.drop(columns=['Exited'])
y=df['Exited']
```

```python
x.shape
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
model=Sequential()
```

```python
model.add(Dense(3, activation='relu', input_dim=9))  # LAYER ONE
model.add(Dense(3, activation='relu'))               # LAYER TWO
model.add(Dense(1, activation='sigmoid'))            # OUTPUT LAYER
```

```python
#                       TO CHECK THE TRAIN-ABLE PARAMETERS 
```

```python
model.summary()
```

```python
#                  COMPILATION OF MODEL
```

```python
model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
#                            FIT THE MODEL
```

```python
model.fit(x_train,y_train,batch_size=32,epochs=10)
```
